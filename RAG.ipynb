## <h2 style="color:blue;"><Strong>What is RAG<h2>

**In natural language processing (NLP), RAG is a type of artificial intelligence (AI) model that combines a *`retrieval component`* with a *`generation component`*. The retrieval component retrieves relevant information from a database or knowledge base, and the generation component uses this information to generate text.**

**Why RAG and not use LLMs directly?**

1. **Token Limits**: RAG chunks input data exceeding LLM's token limit and stores it in a vector database.  
2. **Personal Data**: LLMs cannot correctly answer queries requiring private or custom data.  
3. **Privacy**: RAG ensures control over sensitive or proprietary information.  
4. **Outdated Training**: LLMs may lack updated knowledge; RAG retrieves relevant, current data.  
5. **Cost**: Embedding and vector storage are cheaper than sending large contexts repeatedly.


Retrieval-Augmented Generation (RAG) is a method in natural language processing (NLP) that combines two steps:

- **Retrieval**: The system first searches through an external knowledge base (like a database or a collection of documents) to find relevant information related to a given question or task.

- **Generation**: After retrieving the relevant information, the system uses a generative model (like GPT or T5) to generate a response, combining the retrieved knowledge with the original input.

RAG allows the model to produce more accurate, informed, and contextually relevant outputs by leveraging external data during the generation process.

***Retrieval-Augmented Generation (RAG) Architecture***

The above image represents the architecture of a **Retrieval-Augmented Generation (RAG)** system, which combines both retrieval and generation components. Here's a step-by-step breakdown:

1. **Query Input**: 
   - The process starts with a user query that is fed into the system.

2. **Retrieval Mechanism**: 
   - The query is passed to a retrieval system (commonly using dense vector embeddings or similarity searches). This system scans a large document or knowledge base to find the most relevant information associated with the query.
   - In the architecture, a retriever model like Dense Passage Retrieval (DPR) is typically used. It computes the similarity between the query and documents in a database.

3. **Document Retrieval**: 
   - The retriever returns several relevant documents or passages that contain knowledge relevant to the query. These passages supplement the generative model's knowledge base.

4. **Generative Model (RAG Model)**:
   - The retrieved documents are passed to a generative model (such as GPT or BERT), which processes both the input query and the retrieved documents to generate a coherent and contextually enriched response.
   - This step combines the user's query with the retrieved passages, allowing the model to produce a more accurate and informed answer.

5. **Final Output**:
   - The generative model provides an answer, utilizing both the retrieved knowledge and its own pre-trained knowledge base.

***Key Components:***
- **Retriever and Generator Combination**: The architecture emphasizes that retrieval is performed first, followed by generation, allowing for more accurate and knowledge-based outputs.
- **Multiple Document Retrieval**: Multiple documents are retrieved, enriching the information available to the generative model, which results in better responses.

This combination of retrieval and generation allows the system to provide dynamic, real-time responses by augmenting the generative model with external knowledge sources.

***Challenges:***
1. **Data Preparation**: Cleaning, chunking, and embedding data accurately can be time-consuming.  
2. **Vector Database Setup**: Choosing, setting up, and scaling the vector database (e.g., Pinecone, FAISS).  
3. **Latency**: Additional steps (retrieval, embedding) can increase response times.  
4. **Relevance**: Ensuring the retrieval model fetches the most relevant documents for the query.  
5. **Cost**: Initial embedding generation and database operations can be resource-intensive.  
6. **Fine-Tuning**: Aligning the LLM's output with retrieved context for coherent and accurate answers.  
7. **Maintenance**: Keeping the vector database up-to-date as the knowledge base evolves.  
8. **Integration Complexity**: Combining RAG with existing systems and workflows can require expertise.

| **Feature**                    | **Traditional Databases**                                    | **Vector Databases**                                            |
|---------------------------------|--------------------------------------------------------------|-----------------------------------------------------------------|
| **Data Representation**        | Rows and columns of scalar data (strings, numbers, etc.)     | Vectors (multi-dimensional embeddings) representing data        |
| **Query Type**                  | Exact match queries (e.g., SELECT * WHERE id = ?)            | Similarity search using distance metrics (e.g., cosine similarity, Euclidean distance) |
| **Optimization Focus**          | Optimized for exact matches and indexing based on scalar values | Optimized for approximate nearest neighbor (ANN) search         |
| **Search Method**               | Standard filtering, indexing, and sorting                    | Approximate Nearest Neighbor (ANN) search (e.g., hashing, quantization, graph-based) |
| **Search Accuracy vs Speed**    | Exact match, fast query results                              | Trade-off between accuracy and speed—faster results may be less accurate |
| **Use Cases**                   | Transactional data (banking, CRM, etc.), relational data     | Semantic search, recommendation systems, image and text retrieval |
| **Data Structure**              | Tables with predefined schema                                | Embeddings (vectors) stored in an efficient data structure for fast similarity search |
| **Query Example**               | SELECT * FROM users WHERE name = 'Alice'                     | Find the most similar vectors to a query vector (e.g., find similar documents) |
| **Technology**                  | SQL-based systems (e.g., MySQL, PostgreSQL)                  | Specialized systems (e.g., Pinecone, Weaviate, FAISS, Milvus)   |
| **Accuracy vs Speed Trade-off** | Exact results (no trade-off)                                | Faster queries can lead to approximate results (e.g., lower recall for faster results) |


**Key Challenge**
Traditional scalar-based databases struggle to handle:
1. **High Dimensionality**: Vectors often represent data in hundreds or thousands of dimensions.
2. **Similarity Search**: Real-time queries require specialized algorithms to compute distances or similarities effectively.
3. **Scalability**: Managing and querying large-scale vector datasets can be computationally expensive.

**Steps in Creating Vector Databases**
| **Step**          | **Description**                                                                                 |
|-------------------|-------------------------------------------------------------------------------------------------|
| **Indexing**      | The database indexes vectors using algorithms like Product Quantization (PQ), Locality-Sensitive Hashing (LSH), or Hierarchical Navigable Small World (HNSW) graphs for efficient similarity searching. |
| **Querying**      | A query vector is compared to the indexed dataset vectors to identify nearest neighbors using similarity metrics like cosine similarity or Euclidean distance. |
| **Post-Processing** | Retrieved nearest neighbors may be re-ranked or filtered using additional similarity measures or criteria to refine results. |

### <h3 style="color:red;"><Strong>Vector Embedding<h3>

Vector embeddings, also known as word embeddings or dense vector representations, are a way to represent words, phrases, or documents as numerical vectors in a high-dimensional space. <br>These vectors capture the semantic meaning and context of the words, allowing them to be used in various natural language processing (NLP) tasks, such as text classification, clustering, and information retrieval.

The idea behind vector embeddings is to map words to vectors in a way that:

- 1.Similar words are close together: Words with similar meanings or context are represented by vectors that are close to each other in the vector space.
- 2.Dissimilar words are far apart: Words with different meanings or context are represented by vectors that are far apart in the vector space.

| **Stage**              | **Key Idea**                          | **Example Models**     | **Strength**                       | **Weakness**                          |
|------------------------|---------------------------------------|------------------------|------------------------------------|---------------------------------------|
| **One-Hot Encoding**   | Unique position for each word         | None                   | Simple and intuitive               | No relationships, very large vectors  |
| **Word Embeddings**    | Words close together in meaning       | Word2Vec, GloVe        | Captures similarity and analogy    | Same embedding for all contexts       |
| **Contextual**         | Words depend on sentence context      | ELMo, BERT             | Smart understanding of language    | Computationally expensive             |
| **Multimodal**         | Embeddings for text, images, audio    | CLIP, DALL-E           | Links different data types         | Requires a lot of data and training   |
| **Universal/Fine-Tuned** | Flexible or task-specific embeddings | OpenAI Embeddings      | General-purpose and adaptable      | Fine-tuning can be time-consuming     |


| **Type**              | **Key Idea**                             | **Example Models**          | **Strength**                              | **Weakness**                              |
|-----------------------|------------------------------------------|-----------------------------|------------------------------------------|-------------------------------------------|
| **Auto-Regressive**   | Predicts the next token based on previous tokens | GPT (GPT-4, GPT-3, GPT-2), LLaMA, LaMDA | Great for text generation and completion | Prone to errors in long contexts          |
| **Auto-Encoding**     | Encodes input into a representation for understanding | BERT, RoBERTa, ALBERT, ELECTRA | Excellent for understanding and classification | Not designed for generative tasks         |


To predict the next word we use `Auto Regressive` .Ex `predict the next word`

### <h3 style="color:red;"><Strong>Load the Knowledge Base<h3>

```python
from langchain.document_loaders import TextLoader,DirectoryLoader
```

`glob` is a Python module used to find all the file paths in a directory that match a specified pattern. It stands for **"global"** as it performs pattern matching similar to Unix shell-style wildcards. It's often used for file and directory management tasks.

**Key Features of glob**:  
**Pattern Matching**: It uses patterns with wildcards (`*`, `?`, `[ranges]`) to match filenames or directory names.

- **`*`**: Matches zero or more characters.  
- **`?`**: Matches exactly one character.  
- **`[abc]`**: Matches any one character from the specified set (e.g., `a`, `b`, or `c`).  
- **`[a-z]`**: Matches any one character in the range (e.g., lowercase letters).


```python
import glob   # glob means global and finds
folders=glob.glob("knowledge-base/*")
for folder in folders:  # pattern ("knowledge-base/*") matching (* = one or more)
    print(folder)

mark_down_files=glob.glob("**/*.md", recursive=True)
for file in mark_down_files:
    print(file)
```

```python
# we use directory loader class for  loading pattern matching folders
documents=[]
for folder in folders:
    doc_type=os.path.basename(folder) # Extract the base name (last part) of the folder path.
    # print(doc_type)
    loader = DirectoryLoader(folder,glob="**/*.md",loader_cls=TextLoader,loader_kwargs={"encoding": "utf-8"},show_progress=True)
    
    folder_docs=loader.load()
    
    for doc in folder_docs:
        print(doc.metadata)
        doc.metadata["doc_type"] = doc_type
        print(doc.metadata) # Setting the  the doc_typee  file to the base name in meta data as it is easy to inteprete
        
        documents.append(doc)


```

```python
# documents
# lets get the page_content of company
page_content_company=[ doc.page_content for doc in documents if doc.metadata["doc_type"]=="company"] # easy to filter using base name

for page in page_content_company:
    print(page)
    
# this is the enire content from company folder.U can visualy check if needed
```

### <h3 style="color:red;"><Strong>Chroma db<h3>

**What is Chroma DB?**

Chroma DB is an open-source vector database designed for storing and retrieving high-dimensional embeddings (vectors), commonly used in machine learning and natural language processing tasks. It enables efficient storage, indexing, and querying of vectors generated from models like OpenAI, Hugging Face, and other transformer models.

**How Does Chroma DB Store Data?**

Chroma DB stores vectors (such as embeddings) along with optional metadata, like document IDs, source information, or any additional context. It uses vector indexing techniques to store vectors in a way that allows fast retrieval, even with millions of entries.

1. **Data Representation**:
   - Vectors are typically generated by embedding models and represent semantic information.
   - Chroma DB stores vectors along with their corresponding metadata, such as text source, timestamp, or other attributes.

2. **Storage Engine**:
   - Chroma uses specialized vector storage backends to manage these embeddings, which typically involve approximate nearest neighbor (ANN) search techniques to facilitate fast retrieval.
   - It can use various index types, such as HNSW (Hierarchical Navigable Small World) and other methods for efficient similarity search.

3. **Indexing**:
   - Chroma DB builds indexes on the vectors, allowing you to search through the database for the most similar vectors efficiently. This is done by organizing the vectors into a data structure optimized for similarity queries.

**How Does Chroma DB Retrieve Data?**

Chroma DB uses vector search algorithms (like ANN) to retrieve the most similar vectors to a query vector. When you provide a new vector to Chroma DB for querying, the system searches through the indexed vectors and retrieves the closest ones based on similarity (usually cosine similarity).

1. **Query Vector**:
   - A vector (e.g., a query or text embedding) is passed to Chroma DB for retrieval.
   
2. **Similarity Search**:
   - Chroma performs a search using distance metrics like cosine similarity, Euclidean distance, or inner product. It returns the vectors that are most similar to the input vector based on the metric used.

3. **Return Metadata**:
   - The results will include both the vectors that match the query and any associated metadata, like the source of the embedding (e.g., document name, timestamp).

**When to Use Each Vector Store?**

- Chroma DB is a great choice for developers who want an easy-to-use, flexible, and open-source vector database with fast retrieval.
- FAISS is excellent if you're optimizing for performance and are comfortable with handling more complexity, especially for large-scale applications that need GPU acceleration.
- Pinecone is ideal for applications where you want a fully managed solution that takes care of scaling and maintenance.
- Weaviate is useful if you need to combine vector search with structured data or knowledge graphs.
- Milvus is a good choice if you're dealing with large-scale datasets and need support for multiple indexing techniques.

**Conclusion**

- Chroma DB is a solid, easy-to-use vector store for general-purpose vector search and storage.
- FAISS is more suited for high-performance use cases.
- Pinecone offers a fully managed solution, while Weaviate and Milvus provide more advanced features, including hybrid search and large-scale support.

The choice depends on your specific requirements in terms of scalability, complexity, and functionality.

**Why Chunking is Especially Important in Retrieval-Augmented Generation (RAG)**

Chunking plays a critical role in the context of **Retrieval-Augmented Generation (RAG)** models. RAG models combine two key components:
1. **Retriever**: Finds relevant documents or passages from a large dataset or knowledge base.
2. **Generator**: Uses those documents to generate answers or create content based on the retrieved information.

**Token Limitations**  
RAG models typically rely on large language models (LLMs) that have **token limits**. These models can only process a fixed number of tokens at once, which can be problematic when dealing with long documents or large datasets.

- **Example**: A document that exceeds the token limit of the model can’t be processed in a single pass. Therefore, chunking is necessary to break the document into smaller parts that fit within the model’s token limit.

**Efficient Retrieval**  
In the retrieval phase, it is crucial to break down documents into smaller, meaningful **chunks**. This enables:
- **Better matching**: Chunks can provide more precise and relevant matches when queried by the retriever.
- **Reduced overhead**: Instead of indexing an entire document, you can index smaller chunks, making retrieval faster and more efficient.

For example, if the retriever is searching for information related to "quantum computing," breaking large documents into chunks helps it find the specific part that talks about quantum computing more efficiently than searching the entire document.

**Improved Contextual Relevance**  
When the model retrieves a chunk, it can use that chunk to generate a more **contextually relevant** response. Smaller chunks make it easier for the model to focus on a specific section of information, leading to better output quality.

- **Example**: If the model retrieves an entire document about "machine learning algorithms," it might struggle to generate a focused answer about one specific algorithm. By chunking the document into sections based on algorithm types, it can generate more focused and accurate answers.

**Scaling with Large Datasets**  
RAG systems often deal with vast amounts of data (e.g., millions of documents). Without chunking:
- **Retrieval becomes inefficient**: Searching through large documents without breaking them into chunks can significantly slow down retrieval times.
- **Generation becomes less relevant**: Larger inputs may overwhelm the model, reducing its ability to generate meaningful and coherent responses.

By chunking the documents into smaller, manageable pieces, RAG systems can scale more efficiently, making the process faster and more accurate.

**Overcoming Token Overflows in Generation**  
After the retriever retrieves relevant chunks, the generator (usually an LLM) needs to handle the context. If the retrieved documents are too long:
- **Token overflow** occurs, where the model might not fit the entire context into its input buffer.
- **Chunking** mitigates this by allowing only the most relevant chunks to be passed to the generator, avoiding token overflow and ensuring that the model can focus on the most relevant context.

**Conclusion**  
In **Retrieval-Augmented Generation**, chunking documents into smaller, more focused pieces is crucial for efficient retrieval, better contextual relevance, and working within token limits. By breaking down large documents into manageable chunks, RAG systems can provide more accurate, focused, and efficient responses.

### <h3 style="color:red;"><Strong>RAG Pipeline<h3>

```python
from langchain_openai import ChatOpenAI

model=ChatOpenAI(
    model="gpt-4o", 
    temperature=0,  #Controls the randomness of the model's output. range [0,1] 1 for max randomness
    max_tokens=None,
    timeout=None,
    max_retries=2
)
```

```python
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationalRetrievalChain

# setting up conversation memory for the chat
memory=ConversationBufferMemory(memory_key="chat_history",return_messages=True)

# Use this vector db as as retriver
retriever=faiss_db.as_retriever(search_kwargs={"k":25})

# Creating the chain
chain=ConversationalRetrievalChain.from_llm(llm=model,retriever=retriever,memory=memory)
```

```python
result=chain.invoke("Describe Insurellm in  few Sentences")
result
```

```python
print(result["answer"])
```

### <h3 style="color:red;"><Strong>Using Gradio<h3>

```python
def chat(message,history):
    result=chain.invoke({"question":message})

    return result["answer"]
    
```
